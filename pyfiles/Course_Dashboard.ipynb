# Advanced Machine Learning — Course Dashboard

**M&T Bank — Live Virtual Training**

Use this page to reach every demo and lab notebook for the course. Click a link, it opens directly in Google Colab.

### How to use each notebook
1. Click the link for the notebook you want.
2. In Colab, go to **File → Save a copy in Drive**.
3. Work in your own copy — the original in the shared repo stays untouched.
4. Each notebook has a data-loading line commented out near the top. Your instructor will tell you the URL to paste there before you run the notebook (or it will already be filled in).

### About the Lab notebooks
Each lab has two versions:
- 📝 **Starter** — the exercise, for you to work through.
- ✅ **Solution** — the completed version. It's provided from the start, so try the Starter first before opening it.
